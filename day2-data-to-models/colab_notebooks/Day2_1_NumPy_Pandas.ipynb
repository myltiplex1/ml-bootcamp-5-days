## Part 1 · NumPy

By convention we import NumPy as `np`.

```python
import numpy as np
print("NumPy version:", np.__version__)
```

### 1.1 Why not just use Python lists?
With a list, doing maths on every item needs a loop. With a NumPy array you write the maths **once** and it applies to every element. This is called **vectorisation**.

```python
prices_list = [100, 250, 80]
doubled_list = [p * 2 for p in prices_list]      # needs a loop
print("List:", doubled_list)

prices = np.array([100, 250, 80])
doubled = prices * 2                              # no loop!
print("NumPy:", doubled)
```

Try it on something bigger. We'll time adding 1 to a **million** numbers both ways.

```python
import time

big_list = list(range(1_000_000))
big_array = np.arange(1_000_000)

start = time.time()
result_list = [x + 1 for x in big_list]
list_time = time.time() - start

start = time.time()
result_array = big_array + 1
array_time = time.time() - start

print(f"Python list: {list_time*1000:.1f} ms")
print(f"NumPy array: {array_time*1000:.1f} ms")
print(f"NumPy was about {list_time / array_time:.0f}x faster")
```

### 1.2 Creating arrays

```python
a = np.array([3, 7, 1, 9])          # from a list
z = np.zeros((2, 3))                # 2 rows, 3 columns of zeros
o = np.ones(4)                      # four ones
r = np.arange(0, 10, 2)             # start, stop (not included), step
l = np.linspace(0, 1, 5)            # 5 evenly spaced numbers from 0 to 1

print("a =", a)
print("z =\n", z)
print("o =", o)
print("r =", r)
print("l =", l)
```

### 1.3 Shape and indexing
The **shape** tells you the size of each dimension: `(rows, columns)`. You'll check `.shape` constantly in ML — most errors are shape mismatches!

Indexing starts at **0**. For a 2D array use `[row, column]`.

```python
g = np.array([[1, 2, 3],
              [4, 5, 6]])

print("shape:", g.shape)        # (2, 3)
print("g[1, 2] =", g[1, 2])     # row 1, column 2 -> 6
print("first row:", g[0])       # [1 2 3]
print("last column:", g[:, -1]) # every row, last column -> [3 6]
```

### 1.4 Maths on every element

```python
temps_c = np.array([30, 25, 35])
temps_f = temps_c * 9/5 + 32
print(temps_f)          # [86. 77. 95.]
```

### 1.5 Statistics and filtering
**Boolean filtering** keeps only the elements where a condition is `True`. Pandas uses exactly the same idea to filter rows of a table.

```python
print("mean:", a.mean())
print("max:", a.max())
print("std:", round(a.std(), 2))

print("Which are > 5?", a > 5)      # True/False for each element
print("Keep only > 5:", a[a > 5])   # [7 9]
```

### 1.6 A small real example: exam scores
Rows = students, columns = subjects (Maths, Physics, Chemistry).

`axis=1` means "across the columns" (one answer per student). `axis=0` means "down the rows" (one answer per subject).

```python
scores = np.array([
    [72, 65, 80],   # Ada
    [55, 70, 62],   # Bola
    [90, 85, 88],   # Chidi
    [48, 52, 60],   # Dayo
])
students = ["Ada", "Bola", "Chidi", "Dayo"]

print("Average per student:", scores.mean(axis=1).round(1))
print("Average per subject:", scores.mean(axis=0).round(1))
print("Best student:", students[scores.mean(axis=1).argmax()])
```

### 2.4 Selecting columns
Single brackets → one column (a **Series**). Double brackets → a smaller table (a **DataFrame**).

```python
df["age"].head()
```

```python
df[["sex", "age", "survived"]].head()
```

### 2.5 Filtering rows
The condition inside the brackets gives True/False for every row; only the True rows are kept.

```python
kids = df[df["age"] < 18]
print("Number of children:", len(kids))

women_first_class = df[(df["sex"] == "female") & (df["pclass"] == 1)]
print("Women in 1st class:", len(women_first_class))
```

⚠️ When combining conditions use `&` (and) or `|` (or), and put **brackets around each condition**.

### 2.6 Adding a new column

```python
df["family_size"] = df["sibsp"] + df["parch"] + 1
df[["sibsp", "parch", "family_size"]].head()
```

### 2.7 Group and summarise — the most useful line in EDA
"For each group, compute this." Since `survived` is 0/1, its **mean** is the survival rate.

```python
df.groupby("sex")["survived"].mean().round(2)
```

```python
df.groupby("pclass")["survived"].mean().round(2)
```

### 2.8 Counting and sorting

```python
print(df["embarked"].value_counts())
print()
print(df.sort_values("fare", ascending=False)[["pclass", "sex", "age", "fare"]].head())
```
